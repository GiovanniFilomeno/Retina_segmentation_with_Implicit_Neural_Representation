# Synthetic CPU smoke test

This notebook checks public tensor contracts with generated inputs only. It uses no retinal data or checkpoint, and it is **not a scientific evaluation**. Outputs are intentionally absent from version control.

```python
import retina_inr
from retina_inr.smoke import run_smoke_checks

assert retina_inr.__version__
report = run_smoke_checks()
assert isinstance(report, dict)
```

```python
import torch

from retina_inr import INRSegmentationModel

torch.manual_seed(7)
coords = torch.rand(35, 2, device="cpu") * 2.0 - 1.0
intensities = torch.rand(35, 1, device="cpu")
features = torch.cat((coords, intensities), dim=1)
model = INRSegmentationModel(task="binary", hidden_dim=32, num_layers=2, num_freqs=3).cpu().eval()

with torch.inference_mode():
    logits = model(features)

assert logits.shape == (35, 1)
assert torch.isfinite(logits).all()
```
